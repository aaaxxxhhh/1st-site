```python
import requests
import json

# 서울의 위도와 경도
latitude = 37.5665
longitude = 126.9780

# Open-Meteo API를 사용해 실시간 날씨 가져오기
url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current_weather=true&timezone=Asia/Seoul"

try:
    response = requests.get(url)
    data = response.json()

    if "current_weather" in data:
        current = data["current_weather"]
        temp = current["temperature"]
        windspeed = current["windspeed"]
        weathercode = current["weathercode"]

        # 간단한 날씨 코드 매핑
        weather_status = {
            0: "맑음 ☀️",
            1: "대체로 맑음 🌤️", 2: "구름 조금 ⛅", 3: "흐림 ☁️",
            45: "안개 🌫️", 48: "침적 안개 🌫️",
            51: "가벼운 이슬비 🌧️", 53: "이슬비 🌧️", 55: "강한 이슬비 🌧️",
            61: "약한 비 🌧️", 63: "보통 비 ☔", 65: "강한 비 🌊",
            71: "약한 눈 ❄️", 73: "보통 눈 ❄️", 75: "강한 눈 ☃️",
            77: "싸락눈 🌨️",
            80: "약한 소나기 🌦️", 81: "보통 소나기 🌧️", 82: "강한 소나기 ⛈️",
            85: "약한 눈소나기 🌨️", 86: "강한 눈소나기 ❄️",
            95: "뇌우 ⚡", 96: "우박을 동반한 뇌우 ⛈️", 99: "강한 우박을 동반한 뇌우 ⛈️"
        }

        status = weather_status.get(weathercode, "정보 없음")

        print("======= 실시간 서울 날씨 =======")
        print(f"현재 기온: {temp}°C")
        print(f"날씨 상태: {status}")
        print(f"풍속: {windspeed} km/h")
        print("===============================")
        print("@ahrim♥︎")
    else:
        print("날씨 데이터를 불러오는 데 실패했습니다.")
except Exception as e:
    print(f"에러가 발생했습니다: {e}")
```

```python
# 1. 3가지 상품의 정가(원) 입력받기
onion_price = int(input("양파 정가 입력(원): "))
tomato_price = int(input("토마토 정가 입력(원): "))
potato_price = int(input("감자 정가 입력(원): "))

# 부가세(10%)가 포함된 판매가 계산
onion_total = int(onion_price * 1.1)
tomato_total = int(tomato_price * 1.1)
potato_total = int(potato_price * 1.1)

# 2. 메뉴판 출력
print("\n====================================")
print("       🌿 신선 마켓 메뉴판 🌿")
print("====================================")
print(f" 🧅 양 파   :   {onion_total:,} 원")
print(f" 🍅 토마토   :   {tomato_total:,} 원")
print(f" 🥔 감 자   :   {potato_total:,} 원")
print("------------------------------------")
print("※ 본 메뉴판의 모든 상품 가격은 부가세(10%)가 포함된 최종 금액입니다.")
print("====================================\n")

# 3. 고객으로부터 구매 수량 입력받기
print("🛒 구매하실 수량을 입력해 주세요.")
onion_qty = int(input("🧅 양파 구매 수량(개): ") or 0)
tomato_qty = int(input("🍅 토마토 구매 수량(개): ") or 0)
potato_qty = int(input("🥔 감자 구매 수량(개): ") or 0)

# 4. 금액 계산
onion_sum = onion_total * onion_qty
tomato_sum = tomato_total * tomato_qty
potato_sum = potato_total * potato_qty
grand_total = onion_sum + tomato_sum + potato_sum

# 5. 최종 결제 영수증 출력
print("\n====================================")
print("       🧾 최종 결제 영수증")
print("====================================")
if onion_qty > 0:
    print(f" 🧅 양 파   : {onion_total:,}원 x {onion_qty}개 = {onion_sum:,}원")
if tomato_qty > 0:
    print(f" 🍅 토마토  : {tomato_total:,}원 x {tomato_qty}개 = {tomato_sum:,}원")
if potato_qty > 0:
    print(f" 🥔 감 자   : {potato_total:,}원 x {potato_qty}개 = {potato_sum:,}원")
print("------------------------------------")
print(f" 합계 금액  : {grand_total:,} 원 (부가세 포함)")
print("====================================")
```
