```python
from flask import Flask, jsonify   # 웹 서버(백엔드) 도구
import threading

app = Flask(__name__)

@app.route("/")                 # 프론트엔드: 브라우저에 보낼 화면(HTML)
def home():
    return """
    <h1>헬로 웹!</h1>
    <button onclick="fetch('api/hello').then(r => r.json()).then(d => msg.innerText = d.message)">서버에 물어보기</button>
    <p id="msg"></p>
    """

@app.route("/api/hello")        # 백엔드: 버튼을 누르면 파이썬이 답하는 곳
def hello():
    return jsonify(message="파이썬 서버가 보낸 인사: 안녕!")

threading.Thread(target=lambda: app.run(port=8000), daemon=True).start()   # 서버를 뒤에서 실행

from google.colab import output
output.serve_kernel_port_as_iframe(8000)   # 런타임의 8000번 포트 화면을 결과 칸에 표시
```

```python
exps = [10, 25, 40]          # 몬스터별 경험치
total = 0                    # 합계를 0으로 시작
for e in exps:               # 경험치를 하나씩 꺼내
    total = total + e        # 지금까지의 합계에 더함
print("웨이브 경험치 합계:", total)
print(sum(exps))             # 합계 기능으로 다시 계산
```

```python
# 아래 빈칸에 텍스트 "채널" 대신 전체 데이터프레임 변수를 넣어서 실행해 보세요!
print(df["채널"].value_counts())
print("총 개수:", len(df))
```

```python
df.groupby("채널")["매출"].sum()   # 채널별로 묶어서 매출을 모두 더하기
```

```python
df.groupby("채널")["매출"].sum().sort_values(ascending=False)    # 채널별로 묶어서 매출을 모두 더하기
```

```python
df.groupby("상품")["수량"].sum().sort_values(ascending=False)   # 상품별 수량 합계를 큰 순서로 정렬
```

```python
# 채널별, 요일별 매출 합계를 구하여 출력하고 전체 합계와 일치하는지 검증합니다.
grouped_revenue = df.groupby(['채널', '요일'])['매출'].sum().unstack().fillna(0)
print("=== 채널별 / 요일별 매출 합계 표 ===")
display(grouped_revenue)

print("\n=== 전체 매출 합계 검증 ===")
print("데이터프레임 전체 매출 합계:", df["매출"].sum(), "원")
print("그룹화 데이터 전체 매출 합계:", grouped_revenue.sum().sum(), "원")
```
